````python
from typing import List, Optional
import requests
def get_embeddings(sentences: List[str], max_length: Optional[int] = 256, url: str = "http://192.168.18.237:8025/embeddings_normalize_query") -> List[List[float]]:
    """
    获取句子的嵌入向量。

    :param sentences: 需要编码的句子列表。
    :param max_length: 句子的最大长度（可选，默认为 256）。
    :param url: FastAPI 接口的 URL（可选，默认为本地服务地址）。
    :return: 返回句子的嵌入向量列表。
    """
    # 构造请求体
    data = {
        "content": sentences,
        "max_length": max_length
    }

    try:
        # 发送 POST 请求
        response = requests.post(url, json=data)
        response.raise_for_status()  # 检查请求是否成功
        result = response.json()  # 解析 JSON 响应
        return result["embeddings"]
    except requests.exceptions.RequestException as e:
        print(f"请求失败: {e}")
        return []
    except KeyError:
        print("响应格式错误，未找到 'embeddings' 字段。")
        return []
````

````python
def rename_keys(d, key_mapping):
    """
    递归修改字典中的键名
    :param d: 原始字典
    :param key_mapping: 键名映射字典
    :return: 修改后的字典
    """
    new_dict = {}
    for k, v in d.items():
        # 修改当前键名
        new_key = key_mapping.get(k, k)
        # 如果值是字典，递归处理
        if isinstance(v, dict):
            new_dict[new_key] = rename_keys(v, key_mapping)
        else:
            new_dict[new_key] = v
    return new_dict


# 定义需要修改的键名映射
key_mapping = {
    'COM3105': 'VIEW_COM3105',
    'COM3106': 'VIEW_COM3106',
}

# 修改键名
my_dict = rename_keys(embedding_dict, key_mapping)

print(my_dict.keys())

# 将更新后的嵌套字典保存为新的 .pkl 文件
new_embedding_dict_file = '/Users/rabyte_c/Desktop/experimental/text2sql_renew/vector-US/nasdaq_embedding_dict_updated.pkl'  # 您可以修改为您想要的文件名或路径
with open(new_embedding_dict_file, 'wb') as f:
    pickle.dump(my_dict, f)
````

````python
from util.model_helpers import  async_R1_volcano
import ast
import re


def extract_list_and_text(input_str):
    """
    从输入字符串中提取 Python 列表和解释文字。
    :param input_str: 输入字符串，包含 Markdown 代码块和解释文字
    :return: 提取的列表（Python 对象）和解释文字（字符串）
    """
    # 正则表达式匹配 Python 列表部分
    list_pattern = r'```python\s*(\[.*?\])\s*```'
    list_match = re.search(list_pattern, input_str, re.DOTALL)

    # 提取列表部分
    if list_match:
        list_str = list_match.group(1)
        # 替换 np.float64 为普通浮点数
        cleaned_list_str = re.sub(r'np\.float64\(([^)]+)\)', r'\1', list_str)
        try:
            # 使用 ast.literal_eval 将字符串转换为 Python 列表
            # python_list = eval(cleaned_list_str)
            python_list = ast.literal_eval(list_str)
        except (ValueError, SyntaxError) as e:
            print(f"列表转换失败: {e}")
            python_list = None
    else:
        python_list = None

    # 提取解释文字
    text = re.sub(list_pattern, '', input_str).strip()

    return python_list, text
def generate_product_slang_extraction_prompt(query):
    """
    生成用于抽取行业产品俗称的 prompt。

    参数:
        query (str): 用户输入的查询文本，例如 "2025年中国小家电销售数据"

    返回:
        str: 完整的 prompt，用于让模型识别 query 中的产品俗称并返回对应产品名称列表
    """
    prompt = f"""
请根据用户输入的 query，识别其中涉及的行业产品俗称，并将这些俗称对应的具体产品名称以 Python 列表的形式输出并解释原因。请确保输出的列表只包含产品名称，不包含其他信息。

示例：
1. 如果用户输入的 query 是 "这个小钢炮的油耗怎么样？"，输出应该是 `['小型SUV']`（假设 "小钢炮" 是小型SUV的俗称）。
2. 如果用户输入的 query 是 "彩电销售额"，输出应该是 `['电视', '智能电视']`（假设 "彩电" 是有色电视的俗称）。
3. 如果用户输入的 query 是 "小家电销售数据"，输出应该是 `['空气炸锅', '扫地机器人', '电动牙刷', '美容仪']`。
4. 如果用户输入的 query 是 "软装风格流行趋势"，输出应该是 ['窗帘', '沙发', '地毯', '装饰画']（假设 "软装" 是窗帘、沙发、地毯、装饰画的统称）。
5. 如果用户输入的 query 是 "硬装材料价格"，输出应该是 ['瓷砖', '地板', '涂料', '灯具']（假设 "硬装材料" 是瓷砖、地板、涂料、灯具的统称）。
6. 如果用户输入的 query 是 "数码配件销量"，输出应该是 ['耳机', '充电器', '数据线', '移动电源']（假设 "数码配件" 是耳机、充电器、数据线、移动电源的统称）。
7. 如果用户输入的 query 是 "日本汽车厂商营收数据"，输出应该是 ['丰田', '马自达', '本田', '日产']（假设 "日本汽车厂商" 是丰田、马自达、本田、日产的统称）。

结果请严格按照以下格式输出：
解释:xxx
```python
[产品名称1, 产品名称2, ...]

如果没有识别到任何直接的产品俗称，请先解释愿意再输出一个空列表 []。


用户query如下：\n
{query}
"""
    return prompt
query = '"美容护肤品销量'
product_res = generate_product_slang_extraction_prompt(query)
# Notebook中运行异步代码的正确方式
import asyncio

async def main():
    res = await async_R1_volcano(product_res, "ep-20250204184841-kxvmq")
    return res

# 在Jupyter中这样运行
ress = await main()
res_list, explain = extract_list_and_text(ress)
print(res_list)
````

````python
import openai
from fastapi import HTTPException
import logging
import aiohttp

async def get_gpt_4_1_response(
        prompt: str,
        endpoint: str = "gpt-4.1-2025-04-14"
) -> dict:
    """
    异步调用Gemini 2.5金融投研API获取响应和元数据
    参数:
        prompt (str): 用户查询语句
        endpoint (str): 模型端点 (默认: gpt-4.1-2025-04-14)
        api_key (str): 可选API密钥 (默认使用环境变量)
        base_url (str): 可选API基础URL
    返回:
        dict: 包含响应内容、token使用量和推理过程的字典
    """
    # 配置默认参数
    api_key =  'sk-fZy6ocIFW0c_SDsdDyITwX9WE5qdebst0RtjYacBk1T3BlbkFJLy7OXZrQcCxv33vYHNdzVnnVrSuAT8ixWbH5o5NSEA'
    base_url =  "https://api.openai.com/v1"

    logging.info(f"开始调用gpt 4.1 API (模型: {endpoint})")

    # 构建消息
    messages = [
        {"role": "system",
         "content": "你是讯兔科技研发的金融投研数据查询助手，尤其擅长基于数据库获取所需信息，例如使用NL2SQL方法。"},
        {"role": "user", "content": prompt}
    ]

    # 异步请求
    async with aiohttp.ClientSession() as session:
        headers = {
            "Authorization": f"Bearer {api_key}",
            "Content-Type": "application/json"
        }
        payload = {
            "model": endpoint,
            "messages": messages,
            "extra_params": {
                "return_reasoning": True  # 请求返回推理过程
            }
        }

        try:
            async with session.post(
                    f"{base_url}/chat/completions",
                    headers=headers,
                    json=payload,
                    timeout=aiohttp.ClientTimeout(total=60)
            ) as response:

                if response.status == 200:
                    data = await response.json()
                    result = {
                        "content": data['choices'][0]['message']['content'],
                        "usage": data.get('usage', {}),
                        "reasoning": data['choices'][0]['message'].get('reasoning', '')
                    }

                    # 记录token使用情况
                    logging.info(
                        f"API调用成功 | "
                        f"Prompt tokens: {result['usage'].get('prompt_tokens', 0)} | "
                        f"Completion tokens: {result['usage'].get('completion_tokens', 0)}"
                    )
                    return result['content']

                else:
                    error_msg = await response.text()
                    logging.error(f"API请求失败: {response.status} - {error_msg}")
                    raise Exception(f"API请求失败: {response.status}")

        except asyncio.TimeoutError:
            logging.error("API请求超时")
            raise Exception("请求超时，请稍后重试")
        except Exception as e:
            logging.error(f"API调用异常: {str(e)}")
            raise


# Example usage:
response = await get_gpt_4_1_response("在2022年，电子行业的平均股息率如何？")
print(response)
````
